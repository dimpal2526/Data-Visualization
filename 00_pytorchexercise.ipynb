```python
#Speaking of random seeds, we saw how to set it with torch.manual_seed() but is there a GPU equivalent? (hint: you'll need to look into the do
# set cpu random seed
torch.manual_seed(0)

# if using cuda , set gpu seed too
if torch.cuda.is_available():
  torch.cuda.manual_seed(0)
```

```python
#Create two random tensors of shape (2, 3) and send them both to the GPU (you'll need access to a GPU for this). Set torch.manual_seed(1234) when creating the tensors (this doesn't have to be the GPU random seed)

torch.manual_seed(1234)

tensor_a = torch.rand(2, 3)
tensor_b = torch.rand(2, 3)

#send to gpu if available
device = torch.device("cuda" if torch.cuda.is_available()
else"cpu")

tensor_a = tensor_a.to(device)
tensor_b = tensor_b.to(device)
```

```python
#Perform a matrix multiplication on the tensors you created in 6 (again, you may have to adjust the shapes of one of the tensors).
tensor_b_t = tensor_b.T
result = torch.matmul(tensor_a, tensor_b.T)
print(result)
```

```python
#Find the maximum and minimum values of the output of 7.
#Find the maximum and minimum index values of the output of 7.

min_value = torch.min(result)
max_value = torch.max(result)
print(min_value)
print(max_value)
```
